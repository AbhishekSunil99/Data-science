```python

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
data = np.random.randn(1000)
plt.figure(figsize=(8, 6))
sns.histplot(data, bins=30, kde=False, color='skyblue', edgecolor='black')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Distribution of Random Data using Seaborn')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

'KDE'is mainly used to visualise the overall distribution and concentration of numerical data using a smooth curve.

```python
import matplotlib.pyplot as plt
months=['january','february', 'march', 'april']
sales=[20,30,40,50]
plt.plot(months,sales,marker='^',linestyle="-.")  #[.,*,o,s,p,^-markers]  [linestyle- '-,-.,:,---,--]
plt.xlabel('months')
plt.ylabel('sales')
plt.title('sales per month')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Phone", "Tablet"]
sales = [50, 80, 40]

plt.barh(products, sales)    #ploth - it shows horizontaly  [used to compare categories]

plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product Sales")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.randn(1000)

plt.hist(data, bins=30)    #[to understand the distribution of numerical data]

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Distribution of Data")

plt.show()
```

```python
height = [150, 160, 165, 170, 180]
weight = [50, 55, 60, 65, 75]

plt.scatter(height, weight)    #used to show the relationship between the numerical variables

plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height vs Weight")

plt.show()
```

```python
products = ["Laptop", "Phone", "Tablet"]
sales = [50, 80, 40]

plt.pie(sales, labels=products,autopct="%4.2f%%" )
#autopct is a parameter used to automatically calculate and display percentage values inside each slice of a pie chart

plt.title("Sales Distribution")

plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.fill_between(x, y) #Can be used to show the area under a line.

plt.show()
```

```python
months = ["Jan", "Feb", "Mar"]

online = [100, 150, 180]
offline = [80, 120, 140]

plt.plot(months, online, label="Online")
plt.plot(months, offline, label="Offline")

plt.legend()

plt.show()
```

```python
fig, ax = plt.subplots(1, 2)    #to create multiple graphs in one fig

ax[0].plot([1, 2, 3], [10, 20, 30])

ax[1].bar(["A", "B", "C"], [20, 15, 30])

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Sales": [100, 150, 130, 180]
}

sns.lineplot(x="Month", y="Sales", data=data)

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Sales  Line Plot")

plt.show()
```

```python
data = {
    "Product": ["Laptop", "Phone", "Tablet"],
    "Sales": [50, 80, 40]
}

sns.barplot(x="Product", y="Sales", data=data)

plt.show()
```

```python
data = {
    "Height": [150, 160, 165, 170, 180],
    "Weight": [50, 55, 60, 65, 75]
}

sns.scatterplot(
    x="Height",
    y="Weight",
    data=data
)

plt.show()
```

```python
data = {
    "Height": [150, 160, 165, 170, 180],
    "Weight": [50, 55, 60, 65, 75],
    "Gender": ["F", "M", "F", "M", "F"]
}

sns.scatterplot(
    x="Height",
    y="Weight",
    hue="Gender",
    data=data
)

plt.show()
```

```python
marks = [45, 50, 52, 55, 60, 62, 65, 90]

sns.boxplot(x=marks)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sales = [
    [10, 20, 30, 25],
    [15, 25, 35, 30],
    [20, 30, 40, 35]
]

sns.heatmap(sales, annot=True, cmap="Purples")

plt.show()
```
