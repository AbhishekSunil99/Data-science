```python
import matplotlib.pyplot as plt
months = ["jan","Feb","Mar","Apr","May"]

sales = [20,30,25,40,50]
profit = [5,8,6,12,15]

plt.plot(months,sales,label="sales") #here label means name the label
plt.plot(months,profit,label="profit")
plt.legend() #it means display the names

plt.show()
```

```python
months = ["jan","Feb","Mar","Apr","May"]

sales = [20,30,25,40,50]
plt.plot(months,sales)

plt.xlabel("months")
plt.ylabel("sales")
plt.title("monthly sales")
plt.grid(True)
plt.plot(
    months,
    sales,
    marker="*",  #show individual data points
    color="red",
    linestyle="--"
)
plt.show()
```

```python
months = ["jan","Feb","Mar","Apr","May"]

sales = [20,30,25,40,50]
plt.bar(months,sales)

plt.xlabel("months")                           #bar
plt.ylabel("sales")
plt.title("monthly sales")
plt.grid(True)

plt.show()
```

```python
study_hours = [1,2,3,4,5,6]
marks = [40,45,55,65,75,85]

plt.scatter(study_hours,marks)                  #scatter

plt.title("study_hours vs marks")
plt.xlabel("study_hours")
plt.ylabel("marks")

plt.show()
```

```python
marks = [50,46,50,55,60,40,55,98,
         60,65,70,75,77,44,30,79]

plt.hist(marks)

plt.title("mark ditribution")
plt.xlabel("marks")
plt.ylabel("frequancy")

plt.show()
```

```python
categories = ["Food","Travel","Shopping","Others"]
amount = [4000,2000,3000,1000]

plt.pie(amount,labels=categories)

plt.title("monthly expenses")


plt.show()
```
